## 1. Analisis Data Statistik

Data yang digunakan: `50, 58, 43, 64, 47, 50, 56, 47, 60, 64`.

```python
import math
import numpy as np
import pandas as pd
from scipy import stats

data = np.array([50, 58, 43, 64, 47, 50, 56, 47, 60, 64], dtype=float)
pd.DataFrame({'Observasi': range(1, len(data)+1), 'Nilai': data})
```

### 1.a. Estimasi rata-rata populasi dengan tingkat kepercayaan 99%

Jumlah data adalah `n = 10`. Simpangan baku populasi tidak diketahui, sehingga karena ukuran sampel kecil digunakan distribusi **t-Student**, bukan distribusi z.

Rumus interval kepercayaan:

$$\bar{x} \pm t_{\alpha/2,n-1}\frac{s}{\sqrt{n}}$$

Dengan `α = 0,01` dan derajat bebas `9`.

```python
n = len(data)
mean = data.mean()
s = data.std(ddof=1)
alpha = 0.01
df = n - 1
t_critical = stats.t.ppf(1-alpha/2, df)
margin = t_critical * s / math.sqrt(n)
lower = mean - margin
upper = mean + margin

print(f'Rata-rata = {mean:.2f}')
print(f'Simpangan baku sampel = {s:.4f}')
print(f't kritis = {t_critical:.4f}')
print(f'Margin of error = {margin:.4f}')
print(f'Interval kepercayaan 99% = ({lower:.4f}, {upper:.4f})')
```

### Interpretasi

Interval kepercayaan 99% untuk rata-rata populasi adalah sekitar **46,19 sampai 61,61**. Artinya, berdasarkan sampel yang tersedia, rata-rata populasi diperkirakan berada pada rentang tersebut dengan tingkat kepercayaan 99%.

### 1.b. Standard deviasi

Simpangan baku sampel dihitung dengan:

$$s=\sqrt{\frac{\sum(x_i-\bar{x})^2}{n-1}}$$

```python
kuadrat_deviasi = (data - mean) ** 2
varian = kuadrat_deviasi.sum() / (n-1)
std = math.sqrt(varian)

print(f'Jumlah kuadrat deviasi = {kuadrat_deviasi.sum():.2f}')
print(f'Varian sampel = {varian:.4f}')
print(f'Simpangan baku sampel = {std:.4f}')
```

### 1.c. Skewness dan kurtosis

Skewness menunjukkan kemencengan distribusi, sedangkan kurtosis menunjukkan tingkat keruncingan distribusi. Kode berikut menggunakan **excess kurtosis**, sehingga distribusi normal memiliki nilai sekitar 0.

```python
skewness = stats.skew(data)
kurtosis = stats.kurtosis(data)

print(f'Skewness = {skewness:.6f}')
print(f'Excess kurtosis = {kurtosis:.6f}')
```

### Interpretasi

Skewness sekitar **0,0818**, sehingga distribusi data relatif simetris dengan sedikit kemencengan ke kanan. Excess kurtosis sekitar **-1,3806**, sehingga distribusi cenderung lebih datar dibandingkan distribusi normal.

## 2. K-NN Regression

Tujuan soal adalah menentukan nilai `y` pada data ke-13 berdasarkan `x1 = 14` dan `x2 = 19`.

Soal yang tertulis tidak mencantumkan nilai `k`, sehingga pada perhitungan ini digunakan **k = 3**. Jika nilai `k` dari dosen berbeda, hasil prediksi juga akan berubah.

```python
knn = pd.DataFrame({
    'data': range(1,14),
    'x1': [9,8,9,6,12,5,14,12,3,15,15,2,14],
    'x2': [11,10,13,11,15,10,16,17,7,17,17,4,19],
    'y': [31,28,35,28,42,25,46,46,17,49,49,10,np.nan]
})
knn
```

### Langkah K-NN

Jarak Euclidean dihitung menggunakan:

$$d=\sqrt{(x_1-x_{1t})^2+(x_2-x_{2t})^2}$$

Data kemudian diurutkan berdasarkan jarak terkecil. Untuk `k = 3`, tiga data terdekat digunakan sebagai tetangga dan nilai `y` diprediksi menggunakan rata-ratanya.

```python
target_x1, target_x2 = 14, 19
k = 3
train = knn[knn['y'].notna()].copy()
train['jarak'] = np.sqrt((train['x1']-target_x1)**2 + (train['x2']-target_x2)**2)
urut = train.sort_values('jarak').reset_index(drop=True)
tetangga = urut.head(k)
prediksi = tetangga['y'].mean()

print('Tiga tetangga terdekat:')
display(tetangga[['data','x1','x2','y','jarak']])
print(f'Prediksi y = {prediksi:.2f}')
```

### Hasil K-NN Regression

Tiga tetangga terdekat adalah data ke-10, ke-11, dan ke-8 dengan nilai `y` masing-masing **49, 49, dan 46**.

$$\hat{y}=\frac{49+49+46}{3}=48$$

**Jadi, nilai `y` yang diprediksi untuk data ke-13 adalah 48.**

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.scatter(train['x1'], train['x2'], s=70, label='Data latih')
plt.scatter(tetangga['x1'], tetangga['x2'], s=120, facecolors='none', edgecolors='black', linewidths=2, label='3 tetangga terdekat')
plt.scatter(target_x1, target_x2, s=120, marker='X', label='Data target')
plt.xlabel('x1'); plt.ylabel('x2'); plt.title('Visualisasi K-NN Regression')
plt.grid(alpha=0.25); plt.legend(); plt.show()
```

## 3. Selang Kepercayaan 95% untuk Rata-rata IP Mahasiswa

Diketahui `n = 36`, rata-rata sampel `x̄ = 2,6`, dan simpangan baku **populasi diketahui** `σ = 0,3`. Karena `σ` diketahui, digunakan distribusi normal dengan `z = 1,96`.

Rumus:

$$\bar{x}\pm z_{\alpha/2}\frac{\sigma}{\sqrt{n}}$$

```python
mean_ip = 2.6
sigma = 0.3
n_ip = 36
confidence = 0.95
alpha = 1-confidence
z = stats.norm.ppf(1-alpha/2)
margin_ip = z*sigma/math.sqrt(n_ip)
lower_ip = mean_ip-margin_ip
upper_ip = mean_ip+margin_ip

print(f'z kritis = {z:.2f}')
print(f'Margin of error = {margin_ip:.4f}')
print(f'Interval kepercayaan 95% = ({lower_ip:.4f}, {upper_ip:.4f})')
```

### Kesimpulan

Dengan tingkat kepercayaan 95%, rata-rata IP populasi diperkirakan berada pada interval **2,502 sampai 2,698**, atau sekitar **2,50 sampai 2,70** jika dibulatkan dua angka desimal.

Perbaikan utama pada notebook ini adalah pemilihan metode yang disesuaikan dengan kondisi data: **distribusi t** untuk sampel kecil dengan simpangan baku populasi tidak diketahui, **K-NN regression** untuk prediksi `y`, dan **distribusi z** ketika simpangan baku populasi diketahui.
